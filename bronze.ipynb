```python
import os

import pandas as pd
from sqlalchemy import create_engine

DATABASE_URL = os.getenv("DATABASE_URL")

engine = create_engine(DATABASE_URL)
```

```python
TABLES = [
    "customers",
    "delivery_events",
    "drivers",
    "routes",
    "shipments",
    "vehicles",
    "warehouses",
]

for name in TABLES:
    df = pd.read_csv(f"messy_logistics_data/{name}.csv")
    df.to_sql(name, engine, schema="bronze", if_exists="replace", index=False)
    print(f"Loaded bronze.{name}: {len(df):,} rows")
```
