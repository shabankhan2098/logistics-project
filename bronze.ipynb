```python
import pandas as pd
from sqlalchemy import create_engine, text

engine = create_engine(
    "postgresql+psycopg://postgres:2098@localhost:5432/logistics_project"
)
```

```python
TABLES = [
    "customers",
    "delivery_events",
    "drivers",
    "routes",
    "shipments",
    "vehicles",
    "warehouses",
]

for name in TABLES:
    df = pd.read_csv(f"messy_logistics_data/{name}.csv")
    df.to_sql(
        name,
        con=engine,
        schema="bronze",
        if_exists="replace",
        index=False,
    )
    print(f"Loaded bronze.{name}: {len(df):,} rows")
```
